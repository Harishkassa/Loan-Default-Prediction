#### Min and Max values in each columns

```python
# function to display min and max values of all numeric columns
def show_min_max(df):
    for col in df.select_dtypes(include=['float64', 'int64']).columns:
        print(f"{col}: Min = {df[col].min()}, Max = {df[col].max()}")
```

```python
show_min_max(df)
```

## Clean dataframe

```python
# function to clean dataframe and show before/after samples
def clean_dataframe(df):
    print("Before cleaning...\n")

    # show initial values of each column
    for col in df.columns:
        print(df[col].head())
        print()

    # clean categorical columns (string formatting)
    for col in df.columns:
        if df[col].dtype == 'O':
            df[col] = df[col].str.capitalize().str.replace(" ", "_")

    print("\nAfter cleaning...\n")

    # show cleaned values of each column
    for col in df.columns:
        print(df[col].head())
        print()
```

```python
print("Converting the Upper case object column values into Capitalize\n")

clean_dataframe(df)
```

Observation : Employment_length has all integer values with float type (5 -> 5.0) so we will convert it into integer
